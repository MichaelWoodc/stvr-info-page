```python
# =========================
# Complaint File Index Generator
# =========================

from pathlib import Path
from datetime import datetime

ROOT = Path(".")
OUTPUT_FILE = "complaint_index.html"

EXTENSIONS = {".pdf", ".png", ".jpg", ".jpeg", ".gif", ".webp"}

def build_tree(path, level=0):
    html = ""
    indent = level * 20

    items = sorted(path.iterdir(), key=lambda x: (x.is_file(), x.name.lower()))

    for item in items:
        rel_path = item.relative_to(ROOT)

        if item.is_dir():
            html += f"""
            <div style="margin-left:{indent}px; margin-top:12px;">
                📁 <strong>{item.name}</strong>
            </div>
            """
            html += build_tree(item, level + 1)

        elif item.suffix.lower() in EXTENSIONS:
            html += f"""
            <div class="file-row" style="margin-left:{indent}px;">
                <div class="file-status status-na" contenteditable="true">N/A</div>
                <div class="file-name">
                    📄 <a href="{rel_path.as_posix()}" target="_blank">{item.name}</a>
                </div>
            </div>
            """

    return html

tree_html = build_tree(ROOT)
timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

html = f"""
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<title>Complaint File Index</title>
<style>
body {{
  background-color: #000;
  color: #eee;
  font-family: "Segoe UI", Arial, sans-serif;
  padding: 40px;
}}

h1 {{
  color: #4da3ff;
  text-shadow: 0 0 10px #0044ff66;
}}

a {{
  color: #4da3ff;
  text-decoration: none;
}}

a:hover {{
  text-shadow: 0 0 8px #4da3ff;
}}

.file-row {{
  display: flex;
  align-items: center;
  gap: 16px;
  padding: 4px 0;
}}

.file-status {{
  width: 140px;
  font-weight: bold;
  text-align: center;
  border-radius: 6px;
  padding: 2px 6px;
}}

/* Status colors */
.status-na {{
  background-color: #660000;
  color: #ffb3b3;
}}

.status-waiting {{
  background-color: #660000;
  color: #ffb3b3;
}}

.status-unresolved {{
  background-color: #660000;
  color: #ffb3b3;
}}

.status-response {{
  background-color: #665500;
  color: #fff3b0;
}}

.status-resolved {{
  background-color: #004d00;
  color: #b6ffb6;
}}
</style>

<script>
function updateStatusColors() {{
  document.querySelectorAll('.file-status').forEach(el => {{
    const text = el.innerText.trim().toLowerCase();

    el.classList.remove(
      'status-na',
      'status-waiting',
      'status-unresolved',
      'status-response',
      'status-resolved'
    );

    if (text === 'n/a') el.classList.add('status-na');
    else if (text === 'waiting') el.classList.add('status-waiting');
    else if (text === 'unresolved') el.classList.add('status-unresolved');
    else if (text === 'response received') el.classList.add('status-response');
    else if (text === 'resolved') el.classList.add('status-resolved');
  }});
}}

document.addEventListener("input", updateStatusColors);
document.addEventListener("DOMContentLoaded", updateStatusColors);
</script>

</head>
<body>

<h1>Complaint File Index</h1>
<p>Generated: {timestamp}</p>

{tree_html}

</body>
</html>
"""

with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    f.write(html)

print(f"Index created: {OUTPUT_FILE}")
```

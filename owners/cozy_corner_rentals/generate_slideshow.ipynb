```python
import os
import html

# Path to your images
IMAGE_DIR = r"complaints/bullet_damage"

# Output HTML file
OUTPUT_FILE = "bullet_damage.html"

# Allowed image extensions
EXTS = {".png", ".jpg", ".jpeg", ".gif", ".webp"}

# Collect images
images = [
    f for f in os.listdir(IMAGE_DIR)
    if os.path.splitext(f)[1].lower() in EXTS
]
images.sort()

# Build JS arrays
image_list_js = "[" + ",".join(f'"{html.escape(f)}"' for f in images) + "]"
caption_list_js = "[" + ",".join(f'"{html.escape(os.path.splitext(f)[0])}"' for f in images) + "]"

# Build HTML
html_output = f"""
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<title>Issues Slideshow</title>
<style>
body {{
  background:#000;
  color:#eee;
  font-family: Segoe UI, Arial, sans-serif;
  margin:0;
  padding:20px;
}}

.slideshow {{
  text-align:center;
}}

.slideshow img {{
  width:100%;
  max-height:400px;
  object-fit:contain;
  border-radius:10px;
  box-shadow:0 0 15px rgba(0,0,0,0.7);
}}

button {{
  background:#111;
  color:#4da3ff;
  border:1px solid #4da3ff55;
  padding:10px 20px;
  margin:10px;
  border-radius:6px;
  cursor:pointer;
  font-size:1em;
}}

button:hover {{
  background:#4da3ff22;
}}

.caption {{
  margin-top:10px;
  color:#ccc;
}}
</style>
</head>

<body>

<div class="slideshow">
  <img id="slideImage" src="{IMAGE_DIR}/{images[0]}" alt="Slideshow image">
  <div class="caption" id="slideCaption">{os.path.splitext(images[0])[0]}</div>

  <div>
    <button onclick="prevSlide()">⟵ Previous</button>
    <button onclick="nextSlide()">Next ⟶</button>
  </div>
</div>

<script>
const images = {image_list_js};
const captions = {caption_list_js};
let index = 0;

function updateSlide() {{
    document.getElementById("slideImage").src = "{IMAGE_DIR}/" + images[index];
    document.getElementById("slideCaption").textContent = captions[index];
}}

function nextSlide() {{
    index = (index + 1) % images.length;
    updateSlide();
}}

function prevSlide() {{
    index = (index - 1 + images.length) % images.length;
    updateSlide();
}}
</script>

</body>
</html>
"""

# Write file
with open(OUTPUT_FILE, "w", encoding="utf-8") as f:
    f.write(html_output)

print(f"Slideshow HTML generated: {OUTPUT_FILE}")
```
